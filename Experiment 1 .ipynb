```python

import pandas as pd

# Step 1: Create training dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm', 'Cold'],
    'Humidity': ['Normal', 'High', 'High', 'High', 'Normal'],
    'Wind': ['Strong', 'Strong', 'Weak', 'Strong', 'Weak'],
    'Water': ['Warm', 'Warm', 'Cold', 'Warm', 'Warm'],
    'Forecast': ['Same', 'Same', 'Change', 'Same', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)
print("TRAINING DATASET")
print(df)

# Step 2: Initialize hypothesis
attributes = list(df.columns[:-1])
target = 'EnjoySport'
hypothesis = ['0'] * len(attributes)

# Step 3: Apply FIND-S algorithm
print("\nSTEP-BY-STEP HYPOTHESIS")

for index, row in df.iterrows():
    if row[target] == 'Yes':
        print("\nPositive example:", index + 1)

        if hypothesis == ['0'] * len(attributes):
            hypothesis = list(row[attributes])
        else:
            for i in range(len(attributes)):
                if hypothesis[i] != row[attributes[i]]:
                    hypothesis[i] = '?'

        print("Updated hypothesis:", hypothesis)

    else:
        print("\nNegative example:", index + 1)
        print("Hypothesis unchanged:", hypothesis)

# Step 4: Display final hypothesis
print("\nFINAL MOST SPECIFIC HYPOTHESIS")
for attribute, value in zip(attributes, hypothesis):
    print(attribute, "=", value)

# Step 5: Classify a new sample
sample = ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same']

result = all(
    h == '?' or h == x
    for h, x in zip(hypothesis, sample)
)

print("\nNew sample:", sample)
print("Prediction:", "Enjoy Sport" if result else "Do Not Enjoy Sport")
```
