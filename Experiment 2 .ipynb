```python

import pandas as pd
from itertools import product

# Step 1: Create dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Weak', 'Weak'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
print("TRAINING DATA")
print(df)

X = df.iloc[:, :-1].values.tolist()
y = df.iloc[:, -1].tolist()

n = len(X[0])
S = ['0'] * n
G = [['?'] * n]

# Step 2: Generalization function
def covers(h, x):
    return all(a == '?' or a == b for a, b in zip(h, x))

# Step 3: Process training examples
for index, (x, label) in enumerate(zip(X, y)):
    print("\nExample", index + 1, ":", x, label)

    if label == 'Yes':
        if S == ['0'] * n:
            S = x.copy()
        else:
            for i in range(n):
                if S[i] != x[i]:
                    S[i] = '?'

        G = [g for g in G if covers(g, x)]

    else:
        new_G = []

        for g in G:
            if not covers(g, x):
                new_G.append(g)
                continue

            for i in range(n):
                if S[i] != '?' and S[i] != '0' and S[i] != x[i]:
                    h = g.copy()
                    h[i] = S[i]

                    if covers(h, x):
                        continue

                    if not any(
                        all(a == '?' or a == b for a, b in zip(old, h))
                        for old in new_G
                    ):
                        new_G.append(h)

        G = new_G

    print("Specific boundary S:", S)
    print("General boundary G:", G)

# Step 4: Display results
print("\nFINAL SPECIFIC BOUNDARY")
print(S)

print("\nFINAL GENERAL BOUNDARY")
for hypothesis in G:
    print(hypothesis)

print("\nNote: This is a simplified boundary implementation.")
print("Full Candidate-Elimination requires maintaining all")
print("minimal specializations and removing redundant hypotheses.")

```
